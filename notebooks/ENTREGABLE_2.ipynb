```sql
USE tienda;
```

```sql

-- Ejercicio_1

SELECT nombre, precio 
FROM productos 
ORDER BY precio DESC;
```

```sql
-- Ejercicio_2

SELECT * 
FROM clientes 
WHERE pais IN ('ES', 'MX') 
    AND nombre LIKE 'M%';
```

```sql
-- Ejercicio_3

SELECT * 
FROM productos 
WHERE precio BETWEEN 20 AND 60 
  AND stock > 0;
```

```sql
-- Ejercicio_4

SELECT nombre, COALESCE(telefono, 'sin telefono') AS telefono 
FROM clientes 
WHERE telefono IS NULL;
```

```sql
-- Ejercicio_5

SELECT nombre, precio, (precio * 1.21) AS precio_con_iva 
FROM productos 
ORDER BY precio DESC 
LIMIT 10;
```

```sql
-- Ejercicio_6

SELECT nombre, precio, 
  CASE 
    WHEN precio < 30 THEN 'barato'
    WHEN precio BETWEEN 30 AND 100 THEN 'medio'
    ELSE 'caro'
  END AS clasificacion
FROM productos;
```

```sql
-- Ejercicio_7

SELECT COUNT(DISTINCT pais) AS total_paises 
FROM clientes;
```
