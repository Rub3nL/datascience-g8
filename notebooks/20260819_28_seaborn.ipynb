```python
import seaborn as sns
import matplotlib.pyplot as plt


print(sns.__version__)
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
tips['time']
```

```python
sns.displot(data=tips, x='tip')
```

```python
sns.displot(data=tips, x='tip', hue='sex')
plt.show()
```

```python
sns.displot(data=tips, x='tip', hue='sex', kind='kde')
plt.show()
```

```python
sns.displot(data=tips, x='total_bill', bins=20)
plt.show()
```

```python
sns.kdeplot(data=tips, x='total_bill', hue='sex', fill=True)
plt.show()
```

```python
sns.countplot(data=tips, x='day', hue='sex')
plt.show()
```

```python
sns.stripplot(data=tips, x='day', y='tip', hue='sex', dodge=True)
plt.show()
```

```python
sns.boxenplot(data=tips, x='day',y='total_bill')
plt.show()
```

```python
sns.scatterplot(
    data=tips,
    x='total_bill'
    y='sex'
    hue='day',
    style='time',
    size='size',
    palette='pastel'
)
plt.show()
```

```python
sns.jointplot(data=tips, x='total_bill', y='tip', kind='scatter')
plt.show()
```

```python
sns.pairplot(data=tips, vars=['total_bill','tip','size'], hue='sex')
plt.show()
```
