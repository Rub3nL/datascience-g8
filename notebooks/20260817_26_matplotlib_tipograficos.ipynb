```python
from matplotlib import pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 45]

# 1. Propiedades avanzadas dentro de plt.plot()
plt.plot(
    x,
    y,
    marker="o",
    linestyle="-",
    color="b",
    linewidth=2.5,  # Grosor de la línea
    markersize=8,  # Tamaño de los puntos
    label="Tendencia Principal",  # Etiqueta para la leyenda
)

# 2. Etiquetas y Título con personalización de tamaño y fuente
plt.xlabel("Periodo", fontsize=12, fontweight="bold")
plt.ylabel("Valor", fontsize=12, fontweight="bold")
plt.title("Ejemplo de gráfico de línea avanzado", fontsize=14, pad=15)

# 3. Cuadrícula (Grid) para facilitar la lectura de valores
plt.grid(
    True, linestyle="--", alpha=0.6, color="gray"
)  # alpha define la transparencia

# 4. Límites de los ejes (para dar un margen visual)
plt.xlim(0.5, 5.5)  # Margen horizontal
plt.ylim(0, 50)  # Margen vertical empezando desde 0

# 5. Mostrar la leyenda (requiere el parámetro 'label' en plt.plot)
plt.legend(loc="upper left")

# 6. Anotación opcional para resaltar un punto clave (ej: el valor máximo)
plt.annotate(
    "Máximo",
    xy=(5, 45),
    xytext=(4.2, 40),
    arrowprops=dict(facecolor="black", shrink=0.05, width=1, headwidth=6),
)

plt.show()
```

### Grafico de barras

```python
categorias = ['A', 'B', 'C', 'D']
valores = [40 ,70 ,30 ,90]

plt.bar(categorias,valores,color='#BD1738', width=0.20)
plt.xlabel('Categoria')
plt.ylabel('Frecuencia')
plt.title('Gráfico de barras verticales')
plt.show()
```

```python
plt.bar(categorias,valores, width=0.20, color=['red','black','blue','green'], label=categorias )
plt.xlabel('Categoria')
plt.ylabel('Frecuencia')
plt.title('Gráfico de barras verticales')
plt.legend(title='Categorias')
plt.show()
```

```python
plt.barh(categorias,valores,color='green')
plt.xlabel('Frecuencia')
plt.ylabel('Categoria')
plt.title('Gráfico de barras verticales')
plt.show()
```

### Graficos de pastel

```python
frutas = ['Manzana', 'Platanos', 'Peras']
porcentajes = [30, 50, 20]

plt.pie(porcentajes, labels=frutas, autopct='%1.1f%%')
plt.title('Grafico de pastel')
plt.show()
```

### Graficos de dispersion

```python
import numpy as np

x= np.random.rand(50)
y= np.random.rand(50)

plt.scatter(x,y,color='blue')
plt.xlabel('Variable (x)')
plt.ylabel('Variable (y)')
plt.title('Gráfico de dispersión')
plt.show()
```

### Gráfico Histograma
- Reconoce ceoncentracion, dispersion y asimetria. El número de bins importa, muy pocos bins me quita detalle, pero muchos bins me da ruido visual

```python
datos = np.random.randn(1000)
plt.hist(datos, bins=20, color='orange', edgecolor='black')
plt.xlabel('Valor')
plt.ylabel('FRecuencia')
plt.title('Histograma')
plt.show()
```

### GRáfico de Cjas, bigotes o boxplot

```python
data_box = [np.random.normal(0, std, 100) for std in range(1, 4)]

plt.boxplot(data_box, orientation='vertical', patch_artist=True)
plt.xlabel('Grupo')
plt.ylabel('Valor')
plt.title('Gráfico Boxplot')
plt.show()
```
