```python
import matplotlib
print(matplotlib.__version__)
```

```python
import numpy as np

x = np.linspace(0, 5, 10)
y = x ** 2
```

```python
import matplotlib.pyplot as plt
# https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html
```

```python
plt.plot(x, y,'g')
plt.show()
```

```python
plt.plot(x, y,'b')
plt.show()
```

```python
plt.plot(x, y,'ro')
plt.show()
```

```python
plt.plot(x, y,'ro-')
plt.show()
```

```python
plt.plot(x, y, 'g*')
plt.show()
```

```python
plt.plot(x, y, 'gD')
plt.show()
```

```python
plt.plot(x, y, 'gD:')
plt.show()
```

```python
plt.plot(x, y, color='red', marker='o', linestyle='none', markersize=8, markeredgecolor='black')
plt.show()
```

```python
plt.plot(x, y, color='red', marker='o', linestyle='--', markersize=8, markeredgecolor='none')
plt.show()
```

```python
plt.plot(x, y, color='#FF5733', marker='o', linestyle='--', linewidth=2, markersize=8, markeredgecolor='none')
plt.show()
```
