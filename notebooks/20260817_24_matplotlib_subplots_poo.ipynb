```python
import numpy as np
from matplotlib import pyplot as plt
from matplotlib.figure import Figure
```

```python
fig = Figure()
fig = plt.figure()
```

```python
x = np.linspace(0,5,11)
y1 = x ** 2
y2 = x ** 3
```

```python
fig = plt.figure()

ax = fig.add_subplot()
ax.plot(x,y1, 'ro-')
ax.set_title('Funcion cuadratica')
plt.show()
```

```python
fig2 = plt.figure()

ax = fig2.add_subplot()
ax.plot(x,y2, 'bo-')
ax.set_title('Funcion cúbica')
plt.show()
```

```python
dashboard = plt.figure()

axes1 = dashboard.add_axes([0.1, 0.1, 0.8, 0.8])
axes2 = dashboard.add_axes([0.18, 0.58, 0.35, 0.25])

axes1.figure = fig2.
```

```python
def grafica_cuadratica(ax, x, y):
    ax.plot(x,y, 'ro-')
    ax.set_title('Funcion cuadratica')
    ax.grid(True, linestyle='--', axis='both', gapcolor='green', linewidth=1)
    ax.set_xlabel('Input')
    ax.set_ylabel('Output')

def grafica_cubico(ax, x, y):
    ax.plot(x,y1, 'bo-')
    ax.set_title('Funcion cúbica')
    ax.grid(True, linestyle=':', color='red', linewidth=1)

def grafica_pastel(ax, pct, labels):
    ax.pie(pct, label=labels, autopct='%1.1f%%')
    ax.title('Grafico de pastel')


frutas = ['Manzana', 'Platanos', 'Peras']
porcentajes = [30, 50, 20]

dashboard, axes = plt.subplots(1,2,figsize=(12,5))
grafica_cuadratica(axes[0],x,y1)
# grafica_pastel(axes[1],porcentajes,frutas)
grafica_cubico(axes[1],x,y2)

dashboard.suptitle('Dashboard')
plt.tight_layout()
plt.show()
```

```python
import numpy as np

plt.style.use('classic')
x = np.linspace(1,20,100)

dashboard, axes = plt.subplots(2,4, figsize=(16, 8))

axes[0,0].plot(x, np.cos(x))

axes[1,3].plot(x, np.sin(x),'r--', label='f(x) = sin(x)')
axes[1,3].set_title('Onsa senosoidal')
axes[1,3].set_xlabel('Muestra (ms)')
axes[1,3].set_ylabel('Amplitud')
axes[1,3].legend(loc='upper right')
axes[1,3].grid(True, color='blue', linestyle='--',alpha=0.4)

axes[0,2].plot(x, np.tanh(x),'r--', label='f(x) = tanh(x)')
axes[0,2].set_title('Onsa Tangente')
axes[0,2].set_xlabel('Muestra (ms)')
axes[0,2].set_ylabel('Amplitud')
axes[0,2].legend(loc='upper right')
axes[0,2].grid(True, color='orange', linestyle='--',alpha=0.4)

dashboard.tight_layout(h_pad=0.4)
plt.show()
```
