```python
# 7. Analiza `fare` con histograma y boxplot. Calcula los límites de outliers mediante la regla 1.5·IQR y cuenta cuántas observaciones quedan fuera. ¿Los eliminarías? Justifi ca.

fig, axes = plt.subplots(1, 4, figsize=(12, 5))
sns.histplot(data=df, x='fare', bins=20, kde=True, ax=axes[0])
axes[0].set_title('Histograma fare')

sns.boxenplot(data=df, x='fare', ax=axes[1])
axes[1].set_title('Boxplot fare')

q = df[['age','fare']].quantile([0.25, 0.75]).T

q.columns = ['q1','q3']
q['iqr'] = q['q3'] - q['q1']

lower = q['q1'] - 1.5 * q['iqr']
print(lower)

upper = q['q3'] + 1.5 * q['iqr']
print(upper)

df_sin_outlier = df[(df['fare'] > lower['fare']) & (df['fare'] < upper['fare'])]

sns.boxplot(data=df_sin_outlier, x='fare', ax=axes[2])
axes[2].set_title('Boxplot fare sin autlier')

sns.histplot(data=df_sin_outlier, x='fare', bins=20, kde=True, ax=axes[3])
axes[3].set_title('Histograma fare')

plt.show()
```

```python
# 8. Explora las variables categóricas `sex` y `class` con tablas de frecuencia y gráfi cos de barras. ¿Qué grupos dominan el dataset y qué implicación puede tener para las comparaciones?

# df[['sex','class']].value_counts()

for col in ['sex', 'class']:
    sns.countplot(data=df, x=col)
    plt.title(f'Frecuencia de {col}')
    plt.show()
```

```python
# 9. Calcula la tasa global de supervivencia y represéntala gráfi camente. Interpreta la diferencia entre contar supervivientes y calcular una tasa.


count = df['survived'].value_counts()
print(count)

tasa = df['survived'].mean() * 100
print(tasa)

sns.countplot(data=df, x='survived')
```

```python
# 10. Compara la tasa de supervivencia entre mujeres y hombres. Incluye una tabla y un gráfi co. Describe la asociación observada sin usar lenguaje causal.

sex_rate = df.groupby('sex')['survived'].agg(['count','mean'])
sex_rate['mean'] *= 100

print(sex_rate)

sns.barplot(data=df, x='sex', y='survived', errorbar=None)
```

```python
# 11. Compara la tasa de supervivencia entre primera, segunda y tercera clase. ¿Existe un gradiente visible entre las clases?

class_rate = df.groupby('class')['survived'].agg(['count','mean'])
class_rate['mean'] *= 100


print(class_rate)

sns.barplot(data=df, x='class', y='survived', errorbar=None)
```

```python
# 12. Compara la distribución de `age` entre quienes sobrevivieron y quienes no mediante boxplot o violín. Calcula también la mediana por grupo. 
# ¿La edad parece separar claramente a ambos grupos?

age_state = df.groupby('survived')['age'].agg(['count','mean','median','std'])
print(age_state)

sns.boxplot(data=df, x='survived', y='age')
plt.xticks([0,1],['No sobreviviente', 'Sobreviviente'])
plt.show()
```
